Separating the target from the features 

```python
X = df.drop(columns="Sales_KES") 
y = df["Sales_KES"] 
 
print("X shape:", X.shape) 
print("y shape:", y.shape) 
```

Train-Test Split 

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.20, random_state = RANDOM_STATE)

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))
```

Identifying Feature types 

```python
numeric_features = [ 
    "AdvertisingSpend_KES", 
    "StoreVisits", 
    "DiscountPercent" 
] 
 
categorical_features = [ 
    "Region", 
    "CustomerSegment", 
    "ProductCategory" 
] 
```

Leakage safe preprocessing

```python
numeric_pipeline = Pipeline([ 
    ("imputer", SimpleImputer(strategy="median")), 
    ("scaler", StandardScaler()) 
]) 
 
categorical_pipeline = Pipeline([ 
    ("imputer", SimpleImputer(strategy="most_frequent")), 
    ("onehot", OneHotEncoder(handle_unknown="ignore")) 
]) 
 
preprocessor = ColumnTransformer([ 
    ("num", numeric_pipeline, numeric_features), 
    ("cat", categorical_pipeline, categorical_features) 
]) 
```

Context: Why use a pipeline? Learned preprocessing should be fitted only on training data. Keeping 
preprocessing inside the pipeline helps prevent information leakage during validation. 

Building Baselines

```python
mean_baseline = DummyRegressor(strategy="mean") 
median_baseline = DummyRegressor(strategy="median") 
 
mean_baseline.fit(X_train, y_train) 
median_baseline.fit(X_train, y_train) 
 
mean_pred = mean_baseline.predict(X_test) 
median_pred = median_baseline.predict(X_test) 
```

Evaluation Metrics 
• MAE: average absolute prediction error; lower is better. 
• RMSE: penalizes large errors more heavily than MAE; lower is better. 
• R²: how much variation in the target is explained by the model; higher is generally better. 

```python
def evaluate_regression(y_true, y_pred, name): 
    return pd.DataFrame({ 
        "Model": [name], 
        "MAE": [mean_absolute_error(y_true, y_pred)], 
        "RMSE": [np.sqrt(mean_squared_error(y_true, y_pred))], 
        "R2": [r2_score(y_true, y_pred)] 
    }) 
 
mean_results = evaluate_regression(y_test, mean_pred, "Mean Baseline") 
median_results = evaluate_regression(y_test, median_pred, "Median Baseline") 
 
pd.concat([mean_results, median_results], ignore_index=True) 
```

Linear Regression

```python
linear_model = Pipeline([ 
    ("preprocess", preprocessor), 
    ("model", LinearRegression()) 
]) 
 
linear_model.fit(X_train, y_train) 
linear_pred = linear_model.predict(X_test) 
 
linear_results = evaluate_regression( 
    y_test, linear_pred, "Linear Regression" 
) 
linear_results 
```

Interpretation questions:
--Is MAE lower than the baseline? Yes it is!

--Is RMSE lower? Yes it also is

--Is R² higher? Yes it is as well

--Does the model add useful predictive value? Yes!!
An R^2 of approximately 0.87 indicates that the linear model explains about 87% of the variance in the target variable 'Sales_KES', and the substantially reduced error metrics confirm it performs much better than predicting by the mean or median alone.

Ridge Reggression 

```python
ridge_model = Pipeline([ 
    ("preprocess", preprocessor), 
    ("model", Ridge(alpha=1.0)) 
]) 
 
ridge_model.fit(X_train, y_train) 
ridge_pred = ridge_model.predict(X_test) 
 
ridge_results = evaluate_regression( 
    y_test, ridge_pred, "Ridge Regression" 
) 
ridge_results 
```

Context: Ridge is linear regression with regularization. It discourages very large coefficients and can 
improve stability when predictors are noisy or correlated.

Shallow Decission Tree

```python
tree_model = Pipeline([ 
    ("preprocess", preprocessor), 
    ("model", DecisionTreeRegressor( 
        max_depth=4, 
        min_samples_leaf=8, 
        random_state=RANDOM_STATE 
    )) 
]) 
 
tree_model.fit(X_train, y_train) 
tree_pred = tree_model.predict(X_test) 
 
tree_results = evaluate_regression( 
    y_test, tree_pred, "Decision Tree" 
) 
tree_results 
```

Comparing all models 

```python
comparison = pd.concat([ 
    mean_results, 
    median_results, 
    linear_results, 
    ridge_results, 
    tree_results 
], ignore_index=True) 
 
comparison.sort_values("MAE") 
```

Cross-validation on training data only 

```python
scoring = { 
    "MAE": "neg_mean_absolute_error", 
    "RMSE": "neg_root_mean_squared_error", 
    "R2": "r2" 
} 
 
models = { 
    "Linear Regression": linear_model, 
    "Ridge Regression": ridge_model, 
    "Decision Tree": tree_model 
} 
 
rows = [] 
 
for name, model in models.items(): 
    scores = cross_validate( 
        model, X_train, y_train, 
        cv=5, 
        scoring=scoring 
    ) 
 
    rows.append({ 
        "Model": name, 
        "CV_MAE": -scores["test_MAE"].mean(), 
        "CV_RMSE": -scores["test_RMSE"].mean(), 
        "CV_R2": scores["test_R2"].mean() 
    }) 
 
cv_results = pd.DataFrame(rows) 
cv_results 
```

Questions: Which model performs best on average? Are the cross-validation results consistent 
with the test results? Why is training-data cross-validation safer than repeatedly checking the final 
test set? 

-- Ridge Regression performs best on average with the lowest MAE of 15,971.37 and RMSE of 20,103.54 as well as the highest R^2 value of 0.847907

-- The cross-validation results are in fact consistent with the test results. Ridge and Linear Regression rank as the highest perfomers in both splits, with the decision tree performing considerably worse in both splits as well. 

-- Training-data cross-validation is safer than repeatedly checking the final result because it helps prevent test information leaking into the model selection process

Extention challenge: Try Ridge alpha values 0.1, 1, 10, and 100. 

```python
alphas = [0.1, 1.0, 10.0, 100.0]
ridge_results_list = []

for alpha in alphas:
    # Build pipeline with the specified alpha
    ridge_model = Pipeline([
        ("preprocess", preprocessor),
        ("model", Ridge(alpha=alpha))
    ])
    
    # Train and predict
    ridge_model.fit(X_train, y_train)
    ridge_pred = ridge_model.predict(X_test)
    
    # Evaluate and store results
    res = evaluate_regression(y_test, ridge_pred, f"Ridge (alpha={alpha})")
    ridge_results_list.append(res)

# Combine and display the comparison table
ridge_comparison = pd.concat(ridge_results_list, ignore_index=True)
display(ridge_comparison)
```
